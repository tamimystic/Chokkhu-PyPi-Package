# Chokkhu: End-to-End Tabular Pipeline
This notebook demonstrates a complete machine learning pipeline using chokkhu.

```python
import chokkhu as ck
import pandas as pd
import numpy as np

# 1. Load Data
df = pd.DataFrame({
    'feature1': np.random.randn(100),
    'feature2': np.random.randn(100) * 2,
    'category': np.random.choice(['A', 'B', 'C'], 100),
    'target': np.random.choice([0, 1], 100)
})
df.loc[0:5, 'feature1'] = np.nan # Add missing values
print('Data Loaded:')
display(df.head())
```

```python
# 2. EDA (Exploratory Data Analysis)
ck.eda.tabular(df, target_col='target', save_reports=False)
```

```python
# 3. Data Cleaning
df_clean = ck.clean(df, missing='median', outliers='iqr')
```

```python
# 4. Preprocessing
df_proc, state = ck.preprocess(df_clean, target='target', scale='standard', encode='onehot')
```

```python
# 5. Data Splitting
X_train, X_test, y_train, y_test = ck.split(df_proc, target='target', test_size=0.2)
```

```python
# 6. Model Training
model = ck.train('logistic_regression', X_train.values, y_train.values, epochs=100, learning_rate=0.01)
```

```python
# 7. Evaluation
results = ck.evaluate(model, X_test.values, y_test.values)
print('Evaluation Results:', results)
```
